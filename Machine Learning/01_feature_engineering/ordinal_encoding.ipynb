```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('customer.csv')
df.head()
```

```python
# ordinal features are: review, education
# target feature: purchased
df = df.iloc[:, 2:]
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.drop(columns='purchased'), df[['purchased']], test_size=0.3, random_state=42)
```

```python
X_train
```

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
# initialize an ordinalEncoder() object. 
# The way to do that is to define the categories
# If I hadn't specified the order, OrdinalEncoder would have
# given random order to the categories
oe = OrdinalEncoder(categories=[['Poor', 'Average','Good'], ['School', 'UG', 'PG']])
oe.fit(X_train)
X_train = oe.transform(X_train)
```

```python
X_train
```

```python
X_test = oe.transform(X_test)
```

```python
X_test
```

Label Encoding - Encode target variables with value between 0 and n_classes-1

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
```

```python
le.fit(y_train)
y_train = le.transform(y_train)
y_test = le.transform(y_test)
```
